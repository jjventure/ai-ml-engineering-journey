#### Decorators 

Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
### function copy
### closures
### decorators 
```

```python
### function copy

def welcome():
    return "Welcome to the advanced python course."

welcome()
```

```python
wel = welcome
print(wel())
del welcome
print(wel())
```

```python
### Closure function

def main_welcome(msg):

    def sub_welcome_method():
        print("Welcome to the Advanced Python.")
        print(msg)
        print("Please learn these concepts properly.")

    return sub_welcome_method()   # return type of closure function is usually the sub_method of the closure func

main_welcome("Welcome Everyone!")
```

```python
def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to the Advanced Python.")
        func("Welcome everyone to this tutorial!")  # print gets replaced and the msg will get printed simple...
        print("Please learn these concepts properly.")

    return sub_welcome_method()   # return type of closure function is usually the sub_method of the closure func

main_welcome(print)
```

```python
def main_welcome(func, lst):

    def sub_welcome_method():
        print("Welcome to the Advanced Python.")
        print(func(lst)) # func here simply gets replaced by len | so func(lst) --> len(lst) 
        print("Please learn these concepts properly.")

    return sub_welcome_method()   # return type of closure function is usually the sub_method of the closure func
```

```python
main_welcome(len, [1, 2, 3, 4, 5, 6])
```

```python
### Decorator 

def main_welcome(func):

    def sub_welcome_method():
        print("Welcome to the Advanced Python section.")
        func()
        print("Please learn these concepts properly.")

    return sub_welcome_method() 
```

```python
def course_introduction():
    print("This is an advanced python course.")

course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():    # this func is going as a parameter in main_welcome function 
    print("This is an advanced python course.")
```

```python
### Decorator second example 

# closure 

def my_decorator(func):

    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")

    return wrapper()
```

```python
@my_decorator
def say_hello():   # say_hello func goes into my_decorator function as a parameter and rest gets executed.
    print("Hello...")
```

```python
### Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)

        return wrapper 
    return decorator 
```

```python
@repeat(3)
def say_hello():
    print("Hello...")
```

```python
say_hello()
```
