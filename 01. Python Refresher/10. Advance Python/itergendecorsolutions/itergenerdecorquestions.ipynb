### Assignment 1: Custom Iterator

Create a custom iterator class named `Countdown` that takes a number and counts down to zero. Implement the `__iter__` and `__next__` methods. Test the iterator by using it in a for loop.

```python
class Countdown:
    def __init__(self, number):
        self.count = number 

    def __iter__(self):
        """ The object itself is the iterator. """
        return self

    def __next__(self):
        """ iterates over the iterator and gives each element until exhausted... """ 
        if self.count < 0:
            raise StopIteration 
        else:
            current_count = self.count 
            self.count -= 1 
            return current_count  


# Test the iterator by using it in a for loop 
countdown = Countdown(5)
for number in countdown:
    print(number)
```

### Assignment 2: Custom Iterable Class

Create a class named `MyRange` that mimics the behavior of the built-in `range` function. Implement the `__iter__` and `__next__` methods. Test the class by using it in a for loop.

```python
# mimic the built-in range function. 
# range

class MyRange:
    def __init__(self, start, stop, step):

        if step == 0:
            raise ValueError("MyRange() arg 3 must not be zero.")

        self.start = start 
        self.stop = stop 
        self.step = step  

    def __iter__(self):
        return self 

    def __next__(self):
        # if step is positive, only stop iteration when start is >= stop
        if self.step > 0:
            if self.start >= self.stop:
                raise StopIteration 

        # if step is negative, only stop iteration when start <= stop 
        elif self.step < 0:
            if self.start <= self.stop:
                raise StopIteration  

        current = self.start 
        self.start += self.step 
        return current 

# test 
r1 = MyRange(2, 10, 2)    # 2 4 6 8 
r2 = MyRange(10, 2, -2)   # 10 8 6 4 
r3 = MyRange(1, 5, 0)     # ValueError 

for number in r1:
    print(number)

print()

for number in r2:
    print(number)

for number in r3:
    print(number)
```

### Assignment 3: Generator Function

Write a generator function named `fibonacci` that yields the Fibonacci sequence. Test the generator by iterating over it and printing the first 10 Fibonacci numbers.

```python
def fibonacci():
    a, b = 0, 1 

    while True:
        yield a 
        a, b = b, a + b 

gen = fibonacci()

for i in range(11):
    print(next(gen))
```

### Assignment 5: Chaining Generators

Write two generator functions: `even_numbers` that yields even numbers up to a limit, and `squares` that yields the square of each number from another generator. Chain these generators to produce the squares of even numbers up to 20.

```python
# yields even numbers up to a limit 
def even_numbers(limit):
    for num in range(limit + 1):
        if num % 2 == 0:
            yield num 

# yields the square of each number from even_gen generator
def squares(evens_gen):
    for even in evens_gen:
        yield even ** 2 

# test 
evens_gen = even_numbers(20)
squares_gen = squares(evens_gen)

# produce the squares of even numbers up to 20 
for square in squares_gen:
    print(square)
```

### Assignment 6: Simple Decorator

Write a decorator named `time_it` that measures the execution time of a function. Apply this decorator to a function that calculates the factorial of a number.

```python
# time module is specifically designed for measuring execution/perfomance time. 
import time 

def time_it(func):
    """ measures execution time of a function. """
    def wrapper(n):
        # start timer 
        start_time = time.time()

        # somehow call the original function
        result = func(n)

        # end timer 
        end_time = time.time()

        # calculate elapsed time 
        elapsed_time = end_time - start_time 
        print(f"Your program took {elapsed_time} seconds to run.")

        return result 
    return wrapper 

 

@time_it 
def factorial(n):
    result = 1 
    for i in range(1, n+1):
        result *= i
    return result 


answer = factorial(500)
print(answer)
```

### Assignment 7: Decorator with Arguments

Write a decorator named `repeat` that takes an argument `n` and repeats the execution of the decorated function `n` times. Apply this decorator to a function that prints a message.

```python
def repeat(n):  # Layer 1: Repeats the message()
    def decorator(func):
        def wrapper():

            for _ in range(n):
                func()
        return wrapper 
    return decorator 

@repeat(3)
def message():
    print("Hello!")

message()
```

### Assignment 8: Nested Decorators

Write two decorators: `uppercase` that converts the result of a function to uppercase, and `exclaim` that adds an exclamation mark to the result of a function. Apply both decorators to a function that returns a greeting message.

```python
# decorator 1: uppercase - converts the result of a function to uppercase. 
def uppercase(func):
    def wrapper():
        result = func() 
        result = result.upper() 
        return result
    return wrapper 

# exclaim that adds an exclaimation mark to the result of a function.
def exclaim(func):
    def wrapper():
        result = func()
        result = result + "!"
        return result  
    return wrapper 

# apply both decorators to a function that returns a greeting message 
### Important note!: Decorators closest to the function are applied first.
@exclaim
@uppercase
def greet():
    msg = "Welcome user"
    return msg 

greet()
```

### Assignment 9: Class Decorator

Create a class decorator named `singleton` that ensures a class has only one instance. Apply this decorator to a class named `DatabaseConnection` and test it.

```python
def singleton(cls):
    instance = None 

    def wrapper():
        
        nonlocal instance 
        if instance is None:
            instance = cls()
        return instance 

    return wrapper  


@singleton 
class DatabaseConnection:
    pass 

db1 = DatabaseConnection()
db2 = DatabaseConnection()

print(db1 is db2)
```

Create a custom iterator class named `ReverseString` that iterates over a string in reverse. Write a decorator named `uppercase` that converts the string to uppercase before reversing it. Apply the decorator to the `ReverseString` class.

```python
## decorator 
def uppercase(cls):
    ''' Converts the string to uppercase before reversing it. '''

    def wrapper(string):
        uppercased = string.upper()
        return cls(uppercased) 

    return wrapper 



## custom iterator class 
@uppercase 
class ReverseString:
    ''' Iterates over a string in reverse. '''
    def __init__(self, string):
        self.string = string  
        self.index = len(string) - 1  

    def __iter__(self):
        return self 

    def __next__(self):
        if self.index < 0:
            raise StopIteration
        current_index = self.index
        self.index -= 1 
        return self.string[current_index]

obj = ReverseString("hello")

print(next(obj))
print(next(obj))
print(next(obj))
print(next(obj))
print(next(obj))
```

### Assignment 11: Stateful Generators

Write a stateful generator function named `counter` that takes a start value and increments it by 1 each time it is called. Test the generator by iterating over it and printing the first 10 values.

##### What is the difference between a normal function and a generator function?

1. A normal function uses `return` to send back a result and **finish its execution**.

2. A generator function uses `yield` to produce a value and **pause its execution**. It remembers its state and can **resume from where it paused** when the next value is requested.

return → give result → function ends

yield  → give value → function pauses → remembers state → resumes later

```python
#### Stateful generator function 
# The variable 'start' here is state. 
def counter(start):
    while True:
        yield start 
        start += 1 

# test the generator 
gen = counter(1)

for _ in range(10):
    print(next(gen))
```

### Assignment 12: Generator with Exception Handling

Write a generator function named `safe_divide` that takes a list of numbers and yields the division of each number by a given divisor. Implement exception handling within the generator to handle division by zero.

```python
### generator function 
def safe_divide(lst, divisor):   # list of numbers
    # handling exceptions != raising exceptions
    try: 
        for num in lst:
            result = num / divisor 
            yield result
    except ZeroDivisionError:
        print("Denominator should be greater than zero.")
     

gen = safe_divide([10, 20, 30], 5)

# Test 
for val in gen:
    print(val)
```

### Assignment 13: Context Manager Decorator

Write a decorator named `open_file` that manages the opening and closing of a file. Apply this decorator to a function that writes some text to a file.

```python
# decorator | manages the opening and closing of a file 
def open_file(filename, mode):

    def decorator(func):

        def wrapper(*args):

            with open(filename, mode) as file:
                func(file, *args)

        return wrapper
     
    return decorator 

     

# function | write some text to a file 
@open_file("sample1.txt", 'w')
def write_text(file, text:str):
    file.write(text) 

# test 
write_text("Doing this again to understand how much knowledge I gained through chatpgt guidance for this problem...")
```

### Assignment 14: Infinite Iterator

Create an infinite iterator class named `InfiniteCounter` that starts from a given number and increments by 1 indefinitely. Test the iterator by printing the first 10 values generated by it.

```python
# iterator class 
class InfiniteCounter:
    def __init__(self, number):
        self.number = number  

    def __iter__(self):
        return self 

    def __next__(self):
        current_number = self.number 
        self.number += 1 
        return current_number 

# test | print the first 10 values generated by the iterator 
obj = InfiniteCounter(1)
it = iter(obj)

# looping the iterator 
for _ in range(10):
    print(next(it))
```

### Assignment 15: Generator Pipeline

Write three generator functions: `integers` that yields integers from 1 to 10, `doubles` that yields each integer doubled, and `negatives` that yields the negative of each doubled value. Chain these generators to create a pipeline that produces the negative doubled values of integers from 1 to 10.

```python
### 1st generator function | yields integers from 1 to 10 
def integers():
    for i in range(1, 11):
        yield i 

### 2nd generator function | yields each integer doubled 
def doubles(int_gen):
    for each_int in int_gen:
        yield each_int * 2 

### 3rd generator function | yields the negative of each doubled value 
def negatives(doubles_gen):
    for each_double in doubles_gen:
        yield -each_double  

# creating generators 
int_gen = integers()
doubles_gen = doubles(int_gen)
negatives_gen = negatives(doubles_gen)

## test 
for value in negatives_gen:
    print(value)
```
