```python
# Syntax
lambda arguments: expression
```

```python
def addition(a, b):
    return a + b

addition(2, 3)
```

```python
addition = lambda a, b: a + b
type(addition)
print(addition(5, 6))
```

```python
def even(num):
    if num % 2 == 0:
        return True 
    
even(24)
```

```python
even1 = lambda num: num%2==0 
print(type(even1))
print(even1(24))
```

```python
def addition(x, y, z):
    return x + y + z 

addition(12, 13, 14)
```

```python
addition1 = lambda x, y, z: x + y + z
print(type(addition))
print(addition1(12, 13, 14))
```

```python
### map() - applies a function to all items in an iterable 

numbers = [1, 2, 3, 4, 5, 6]

squared_numbers = list(map(lambda x: x ** 2, numbers)) 
print(squared_numbers)
```
