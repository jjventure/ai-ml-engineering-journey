### Assignment 1: Polymorphism with Methods

Create a base class named `Shape` with a method `area`. Create two derived classes `Circle` and `Square` that override the `area` method. Create a list of `Shape` objects and call the `area` method on each object to demonstrate polymorphism.

```python
import math 

# base class 
class Shape:

    def area(self):
        return "Area of the shape" 


# derived class1
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius 

    def area(self):
        return math.pi * (self.radius ** 2)

# derived class2
class Square(Shape):

    def __init__(self, side):
        self.side = side 

    def area(self):
        return self.side ** 2 

# Create a list of shape objects and call the area method on each object to demonstrate polymorphism.
shapes = [Circle(5), Square(4)]

# calling the area method of each object created
for shape in shapes:
    print(shape.area())
```

### Assignment 3: Abstract Base Class with Abstract Methods

Create an abstract base class named `Vehicle` with an abstract method `start_engine`. Create derived classes `Car` and `Bike` that implement the `start_engine` method. Create objects of the derived classes and call the `start_engine` method.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod 
    def start_engine(self):
        pass 

# derived class1 
class Car(Vehicle):
    def start_engine(self):
        print("Car engine is getting started...")

# derived class2 
class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine is getting started...")

# Create objs of the derived classes and call the start_engine() method 
car = Car()
bike = Bike()

car.start_engine()
bike.start_engine()
```

### Assignment 4: Abstract Base Class with Concrete Methods

In the `Vehicle` class, add a concrete method `fuel_type` that returns a generic fuel type. Override this method in `Car` and `Bike` classes to return specific fuel types. Create objects of the derived classes and call the `fuel_type` method.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod 
    def start_engine(self):
        pass 

    def fuel_type(self):
        return f"Generic fuel type."
    
# derived class1 
class Car(Vehicle):
    def start_engine(self):
        print("Car engine is getting started...")

    def fuel_type(self):
        return f"Car runs on petrol or diesel or both."

# derived class2 
class Bike(Vehicle):
    def start_engine(self):
        print("Bike engine is getting started...")

    def fuel_type(self):
        return f"Bike runs on petrol."

# Create objs of the derived classes and call the start_engine() method 
car = Car()
bike = Bike()

# car.start_engine()
# bike.start_engine()

print(car.fuel_type())
print(bike.fuel_type())
```

### Assignment 5: Encapsulation with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Ensure that the balance cannot be accessed directly.

```python
class BankAccount:

    def __init__(self, account_number, balance):
        self.__account_number = account_number 
        self.__balance = balance 

    def deposit(self, amount):
        self.__balance += amount 
        return f"Current Balance: {self.__balance} Rs."

    def withdraw(self, amount):
        if amount < self.__balance:
            self.__balance -= amount 
            return f"Current Balance: {self.__balance} Rs."
        else:
            return "Not enough funds in your account!"

    def check_balance(self):
        return self.__balance 

account = BankAccount(2314104350, 1000)
print(account.check_balance())

dir(account)
```

### Assignment 7: Combining Encapsulation and Inheritance

Create a base class named `Person` with private attributes `name` and `age`. Add methods to get and set these attributes. Create a derived class named `Student` that adds an attribute `student_id`. Create an object of the `Student` class and test the encapsulation.

```python
# base class 
class Person:

    def __init__(self, name, age):
        self.__name = name 
        self.__age = age 

    # getter for name
    def get_name(self):
        return self.__name 

    # getter for age 
    def get_age(self):
        return self.__age 

    # setter for name 
    def set_name(self, name):
        self.__name = name 
        return self.__name 

    # setter for age 
    def set_age(self, age):
        self.__age = age 
        return self.__age 

# derived class 
class Student(Person):

    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

# Create an object of the Student class and test the encapsulation 
student = Student("Joel", 22, 2314104350)

print(student.get_name())
print(student.get_age())

print(student.set_name("Child of God"))
print(student.set_age("God's plan"))
```

### Assignment 9: Abstract Methods in Base Class

Create an abstract base class named `Employee` with an abstract method `calculate_salary`. Create two derived classes `FullTimeEmployee` and `PartTimeEmployee` that implement the `calculate_salary` method. Create objects of the derived classes and call the `calculate_salary` method.

```python
from abc import ABC, abstractmethod

# Abstract base class  
class Employee:
    @abstractmethod 
    def calculate_salary(self):
        pass 


# Derived class1 
class FullTimeEmployee(Employee):
    def calculate_salary(self):
        print("Full time payroll!") 


# Derived class2 
class PartTimeEmployee(Employee):
    def calculate_salary(self):
        print("Parttime payroll...")

# create objects of the derived classes 
full_time_employee = FullTimeEmployee()
part_time_employee = PartTimeEmployee()

# call the calculate_salary method
full_time_employee.calculate_salary()
part_time_employee.calculate_salary()
```

### Assignment 10: Encapsulation in Data Classes

Create a data class named `Product` with private attributes `product_id`, `name`, and `price`. Add methods to get and set these attributes. Ensure that the price cannot be set to a negative value.

```python
class Product:

    def __init__(self, product_id, name, price):
        self.__product_id = product_id 
        self.__name = name 
        self.__price = price 

    # getter for product_id
    def get_product_id(self):
        return self.__product_id

    # getter for name 
    def get_name(self):
        return self.__name 

    # getter for price 
    def get_price(self):
        return self.__price 

    # setter for product_id
    def set_product_id(self, product_id):
        self.__product_id = product_id
        return self.__product_id

    # setter for name 
    def set_name(self, name):
        self.__name = name 
        return self.__name 

    # setter for price  ## price cannot be a negative value
    def set_price(self, price):

        if price > 0:
            self.__price = price 
            return self.__price
        
        else:
            return f"price cannot be a negative value..."

product = Product(1, "RTX 5090", 250000)

print(product.get_product_id())
print(product.get_name())
print(product.get_price())
print()
print(product.set_product_id(2))
print(product.set_name("RTX 5090 Ti"))
print(product.set_price(0))
```

### Assignment 11: Polymorphism with Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:

    def __init__(self, x, y):
        self.x = x 
        self.y = y 

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)

print(v1 + v2)
```

### Assignment 12: Abstract Properties

Create an abstract base class named `Appliance` with an abstract property `power`. Create two derived classes `WashingMachine` and `Refrigerator` that implement the `power` property. Create objects of the derived classes and access the `power` property.

```python
from abc import ABC, abstractmethod

# abstract base class 
class Appliance:
    @abstractmethod 
    def power(self):
        pass 


# Derived class1 
class WashingMachine(Appliance):
    def power(self):
        return "Washing Machine is powering on..."

# Derived class2 
class Refrigerator(Appliance):
    def power(self):
        return "Fridge is powering on..." 


# Create objects of derived classes
washingmachine = WashingMachine()
fridge = Refrigerator()

# access the power property
print(washingmachine.power())
print(fridge.power())
```

### Assignment 13: Encapsulation in Class Hierarchies

Create a base class named `Account` with private attributes `account_number` and `balance`. Add methods to get and set these attributes. Create a derived class named `SavingsAccount` that adds an attribute `interest_rate`. Create an object of the `SavingsAccount` class and test the encapsulation.

```python
# base class 
class Account:

    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance 

    # getter for account_number 
    def get_account_number(self):
        return self.__account_number 

    # getter for balance 
    def get_balance(self):
        return self.__balance

    # setter for account_number 
    def set_account_number(self, account_number):
        self.__account_number = account_number 
        return self.__account_number

    # setter for balance 
    def set_balance(self, balance):
        self.__balance = balance 
        return self.__balance 


# Derived class 
class SavingsAccount(Account):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate 

# create obj of SavingsAccount class 
savingsaccount = SavingsAccount(2314104350, 1000, 7)

# test the encapsulation
print(savingsaccount.get_account_number())
print(savingsaccount.get_balance())
print(savingsaccount.interest_rate)

print()

print(savingsaccount.set_account_number(5000100001))
print(savingsaccount.set_balance(2000))
```

### Assignment 14: Polymorphism with Multiple Inheritance

Create a class named `Flyer` with a method `fly`. Create a class named `Swimmer` with a method `swim`. Create a class named `Superhero` that inherits from both `Flyer` and `Swimmer` and overrides both methods. Create an object of the `Superhero` class and call both methods.

```python
class Flyer:
    def fly(self):
        print("Flyer can only fly...") 

class Swimmer:
    def swim(self):
        print("Swimmer can only swim...") 

class Superhero(Flyer, Swimmer):
    def fly(self):
        print("Superhero can fly!")

    def swim(self):
        print("Superhero can swim!")


# create an obj of Superhero class 
superhero = Superhero()

# call both methods
superhero.fly()
superhero.swim()
```

### Assignment 15: Abstract Methods and Multiple Inheritance

Create an abstract base class named `Worker` with an abstract method `work`. Create two derived classes `Engineer` and `Doctor` that implement the `work` method. Create another derived class `Scientist` that inherits from both `Engineer` and `Doctor`. Create an object of the `Scientist` class and call the `work` method.

```python
from abc import ABC, abstractmethod

# abstract base class 
class Worker:
    @abstractmethod
    def work(self):
        pass 

# Derived class1 
class Engineer(Worker):
    def work(self):
        print("Engineer works for innovation and future tech.") 


# Derived class2 
class Doctor(Worker):
    def work(self):
        print("Doctor provides healthcare services.") 


# Derived class 3
class Scientist(Engineer, Doctor):
    def work(self):
        super().work()
        Doctor.work(self)


# Create an object of the scientist class 
scientist = Scientist()

# Call the work method
scientist.work()
```
