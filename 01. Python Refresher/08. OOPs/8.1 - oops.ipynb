#### Classes and Objects

Object-Oriented Programming (OOP) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson covers the basics of creating classes and objects, including instance variables and methods. 

```python
class Car:
    pass 

# objects 
audi = Car()
bmw = Car()

print(type(audi))
```

```python
print(audi)
print(bmw)
```

```python
# windows is attribute 
audi.windows = 4 
print(audi.windows)
```

```python
tata = Car()
tata.doors = 4 
print(tata.windows)
```

```python
## Instance variable and methods

class Dog:

    # constructor 
    def __init__(self, name, age):
        # instance variables
        self.name = name
        self.age = age 

# Creating Objects 
dog1 = Dog("Buddy", 3)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("Lucy", 4) 
print(dog2.name)
print(dog2.age) 
```

```python
## define a class with instance method 
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age 

    # instance method 
    def bark(self):
        print(f"{self.name} says woof!")

dog1 = Dog("Buddy", 3)
print(dog1.bark())

dog2 = Dog("Lucy", 4)
print(dog2.bark())
```

```python
## Modelling a bank account 

# define a class for bank account 

# always a good practice to use capital letter when defining a class!
class BankAccount:
    def __init__(self, owner, balance = 0):
        self.owner = owner 
        self.balance = balance 

    def deposit(self, amount):
        self.balance += amount 
        print(f"Rs.{amount} is credited.\nNew balance is {self.balance}")

    def withdraw(self, amount):
        # if amount is greater than balance: we cant withdraw...
        if amount > self.balance:
            print("Insufficient funds!")

        else:
            self.balance -= amount 
            print(f"Rs.{amount} is debited.\nNew balance is {self.balance}.")

    def get_balance(self):
        return self.balance 

## create an object 

account = BankAccount("Joel", 5000)
print(account.balance)
```

```python
# Call instance methods
account.deposit(100)
```

```python
account.withdraw(300)
```

```python
print(account.get_balance())
```
