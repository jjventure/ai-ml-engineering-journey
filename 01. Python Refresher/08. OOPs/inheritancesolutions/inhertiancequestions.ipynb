Create a base class named `Animal` with attributes `name` and `species`. Create a derived class named `Dog` that inherits from `Animal` and adds an attribute `breed`. Create an object of the `Dog` class and print its attributes.

```python
class Animal:
    def __init__(self, name, species):
        self.name = name 
        self.species = species 

class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed 

    def __str__(self):
        return f"Dog(Name: {self.name}, Species: {self.species}, Breed: {self.breed})"

dog1 = Dog("Jacky", "Canine", "Golden Retriever")
print(dog1)
```

### Assignment 3: Single Inheritance with Additional Methods

In the `Dog` class, add a method named `bark` that prints a barking sound. Create an object of the class and call the method.

```python
class Animal:
    def __init__(self, name, species):
        self.name = name 
        self.species = species 

class Dog(Animal):
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed 

    def bark(self):
        print("Bhau bhau!")

    def __str__(self):
        return f"Dog(Name: {self.name}, Species: {self.species}, Breed: {self.breed})"

dog1 = Dog("Jacky", "Canine", "Golden Retriever")
print(dog1)

# call the method: bark 
dog1.bark()
```

### Assignment 5: Method Resolution Order (MRO) in Multiple Inheritance

In the `Athlete` class, override the `walk` method to print a different message. Create an object of the class and call the `walk` method. Use the `super()` function to call the `walk` method of the `Walker` class.

```python

# base class 1
class Walker:
    def walk(self):
        print("Walking...")

# base class 2
class Runner:
    def run(self):
        print("Running...")

# derived class: deriving from both Walker and Runner 
class Athlete(Walker, Runner):
    def __init__(self):
        super().__init__()

    def walk(self):
        print('Athlete is walking...')
        super().walk()

athlete1 = Athlete()
athlete1.walk()
# athlete1.run()
```

### Assignment 6: Multiple Inheritance with Additional Attributes

In the `Athlete` class, add an attribute `training_hours` and a method `train` that prints the training hours. Create an object of the class and call the method.

```python

# base class 1
class Walker:
    def walk(self):
        print("Walking...")

# base class 2
class Runner:
    def run(self):
        print("Running...")

# derived class: deriving from both Walker and Runner 
class Athlete(Walker, Runner):
    def __init__(self, training_hours):
        super().__init__()
        self.training_hours = training_hours

    def walk(self):
        print('Athlete is walking...')
        super().walk()

    def train(self):
        print(f"Training for {self.training_hours} hours.")

athlete1 = Athlete(100)
# athlete1.walk()
# athlete1.run()

athlete1.train()
```

### Assignment 7: Diamond Problem in Multiple Inheritance

Create a class named `A` with a method `show` that prints a message. Create two derived classes `B` and `C` that inherit from `A` and override the `show` method. Create a class `D` that inherits from both `B` and `C`. Create an object of the `D` class and call the `show` method. Observe the method resolution order.

```python
# Parent class 
class A:
    def show(self):
        print("A's show method")

# Child class 1
class B(A):
    def show(self):
        print("B's show method") 

# Child class 2 
class C(A):
    def show(self):
        print("C's show method") 

# class D inherits from both B and C 
class D(B, C):
    pass 

d = D()
d.show()
```

### Assignment 8: Using `super()` in Single Inheritance

Create a base class named `Shape` with an attribute `color`. Create a derived class named `Circle` that inherits from `Shape` and adds an attribute `radius`. Use the `super()` function to initialize the attributes. Create an object of the `Circle` class and print its attributes.

```python
# base class 
class Shape:
    def __init__(self, color):
        self.color = color 

# derived class 
class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)
        self.radius = radius 

# create an obj of circle and print its attr 
circle1 = Circle("Red", 3)
print(circle1.color, circle1.radius)
```

### Assignment 9: Using `super()` in Multiple Inheritance

Create a class named `Person` with an attribute `name`. Create a class named `Employee` with an attribute `employee_id`. Create a derived class `Manager` that inherits from both `Person` and `Employee`. Use the `super()` function to initialize the attributes. Create an object of the `Manager` class and print its attributes.

```python
class Person:
    def __init__(self, name):
        self.name = name 

class Employee:
    def __init__(self, employee_id):
        self.employee_id = employee_id

# derived class 
class Manager(Person, Employee):
    def __init__(self, name, employee_id):
        super().__init__(name)
        Employee.__init__(self, employee_id)  # have to pass self manually, coz we are accessing directly through the class and not the object.

# check method resolution order (m-r-o)
# print(Manager.__mro__)

# # create an obj of Manager class and print its attr 
manager1 = Manager("Joel", 22)
print(manager1.name, manager1.employee_id)
```

### Assignment 12: Complex Multiple Inheritance

Create a class named `Base1` with an attribute `a`. Create a class named `Base2` with an attribute `b`. Create a class named `Derived` that inherits from both `Base1` and `Base2` and adds an attribute `c`. Initialize all attributes using the `super()` function. Create an object of the `Derived` class and print its attributes.

```python
class Base1:
    def __init__(self, a, **kwargs):
        self.a = a 
        super().__init__(**kwargs)

class Base2:
    def __init__(self, b, **kwargs):
        self.b = b 
        super().__init__(**kwargs)

# derived class 
class Derived(Base1, Base2):
    def __init__(self, a, b, c):
        self.c = c 
        super().__init__(a = a, b = b)
        

# create an object of the Derived class and print its attributes.
derived1 = Derived("a", "b", "c")
print(derived1.a, derived1.b, derived1.c)
```

### Assignment 13: Checking Instance Types with Inheritance

Create a base class named `Animal` and a derived class named `Cat`. Create objects of both classes and use the `isinstance` function to check the instance types.

```python
# base class 
class Animal:
    def speak(self):
        print("Animal speaks!") 

# derived class
class Cat(Animal):
    def speak(self):
        print("Meow!")

# create objs of both classes and use the isinstance function to check instance types.
animal1 = Animal()
cat1 = Cat()

print(isinstance(animal1, Animal))
print(isinstance(cat1, Animal))
print(isinstance(cat1, Cat))
print(isinstance(animal1, Cat))
```

### Assignment 14: Polymorphism with Inheritance

Create a base class named `Bird` with a method `speak`. Create two derived classes `Parrot` and `Penguin` that override the `speak` method. Create a list of `Bird` objects and call the `speak` method on each object to demonstrate polymorphism.

```python
# base class 
class Bird:
    def speak(self):
        print("Bird Speaks!")

# derived class1
class Parrot(Bird):
    def speak(self):
        print("Parrot says: Squawk!")

# derived class2
class Penguin(Bird):
    def speak(self):
        print("Penguin says: Honk!") 

# Create a list of `Bird` objects 
bird = Bird()
parrot = Parrot()
penguin = Penguin()

birds = [bird, parrot, penguin]

# call the `speak` method on each object to demonstrate polymorphism
for each_bird in birds:
    each_bird.speak()
```

### Assignment 15: Combining Single and Multiple Inheritance

Create a base class named `Device` with an attribute `brand`. Create a derived class `Phone` that inherits from `Device` and adds an attribute `model`. Create another base class `Camera` with an attribute `resolution`. Create a derived class `Smartphone` that inherits from both `Phone` and `Camera`. Create an object of the `Smartphone` class and print its attributes.

```python
# 1st base class
class Device:
    def __init__(self, brand):
        self.brand = brand 

# derived class of Device
class Phone(Device):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model 

# 2nd base class 
class Camera:
    def __init__(self, resolution):
        self.resolution = resolution 

# derived class of Phone and Camera
class Smartphone(Phone, Camera):
    def __init__(self, brand, model, resolution):
        super().__init__(brand, model)
        Camera.__init__(self, resolution)

# Create an obj of Smartphone class 
smartphone1 = Smartphone("Apple", "17 pro max", "max res")

# print its attributes
print(smartphone1.brand, smartphone1.model, smartphone1.resolution)
```
