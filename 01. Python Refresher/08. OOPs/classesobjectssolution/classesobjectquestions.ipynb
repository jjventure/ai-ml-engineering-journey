### Assignment 4: Class with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Create an object of the class and perform some operations.

```python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.__account_number = account_number
        self.__balance = balance 

    # deposit method 
    def deposit(self, amount):
        self.__balance += amount 

    # withdraw money
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Insufficient funds!")

    # check balance 
    def check_balance(self):
        return self.__balance

# create an object and perform operations
acc1 = BankAccount(100010570632, 2000)

# deposited 500 
acc1.deposit(500)

# checking balance after depositing money
print(acc1.check_balance())

# withdrawing 300
acc1.withdraw(300)

# checking balance after withdrawing money
print(acc1.check_balance())
```

### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
from abc import ABC, abstractmethod 
import math 

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass 

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius 

    def area(self):
        return math.pi * (self.radius ** 2)

class Square(Shape):
    def __init__(self, side):
        self.side = side 

    def area(self):
        return self.side * self.side 


circle1 = Circle(5)
square1 = Square(5)

print(circle1.area())
print(square1.area())
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x 
        self.y = y 

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# create objects of the class 
v1 = Vector(1, 2)
v2 = Vector(3, 4)

# test the overloading
v3 = v1 + v2
print(v3)
```

### Assignment 13: Class with Custom Exception

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.

```python
class InsufficientBalanceError(Exception):
    pass 

class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number 
        self.balance = balance 

    def deposit(self, amount):
        self.balance += amount 
        return f"Current balance = {self.balance} Rs."

    def withdraw(self, amount):
        if amount < self.balance:
            self.balance -= amount 
            return f"Current balance = {self.balance} Rs."
        
        else:
            raise InsufficientBalanceError("Not enough funds in your account!")
            
    def check_balance(self):
        return f"Current Balance: {self.balance} Rs."

account1 = BankAccount(2314104350, 1000)

account1.deposit(500)
account1.withdraw(100)
account1.check_balance()
```

### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:

    def __init__(self, filename, filemode):
        self.filename = filename 
        self.filemode = filemode 

    def __enter__(self):
        self.file = open(self.filename, self.filemode)
        return self.file 

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()

# Test
with FileManager('sample.txt', 'r') as file:
    content = file.read()
    print(content)
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:

    def __init__(self, result):
        self.result = result 

    def add(self, num):
        self.result += num 
        return self 

    def subtract(self, num):
        self.result -= num 
        return self 

    def multiply(self, num):
        self.result *= num 
        return self 

    def divide(self, num):
        self.result /= num 
        return self 

calc = Calculator(10)
calc.add(5).multiply(2).subtract(4)
print(calc.result)
```
