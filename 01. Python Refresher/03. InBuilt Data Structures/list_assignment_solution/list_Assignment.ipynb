### Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

transpose = []
# col - stays
for col in range(3):
    new_row = []
    # row 
    for row in range(3):
        new_row.append(matrix[row][col])
    transpose.append(new_row)

print("Original Matrix: ")
for row in matrix:
    print(row)

print("\nTransposed Matrix: ")
for row in transpose:
    print(row)
```

```python
def matrix_transpose(matrix):
    transpose = []
    # col - stays
    for col in range(3):
        new_row = []

        # row 
        for row in range(3):
            new_row.append(matrix[row][col])
        transpose.append(new_row)

    return transpose

matrix_transpose([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
])
```

### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

```python
def flatten_nested_list(nested):
    flattened = []

    for inner_list in range(len(nested)):
        for element in range(len(nested)):
            collected_elements = nested[inner_list][element]
            flattened.append(collected_elements)

    return flattened 

flatten_nested_list([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
])
```

### Assignment 13: List Reversal

Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.

```python
def reverse_list(lst:list):
    reversed_list = lst[::-1]
    return reversed_list

# print("Original List: ")
# print(lst)

print("Reversed List: ")
print(reverse_list([1, 2, 3, 4, 5]))
```

### Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.

```python
def rotate_list(listy:list, n):
    rotated_list = listy[-n: ] + listy[:-n]
    return rotated_list

print(rotate_list([1, 2, 3, 4, 5], 3))
```

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
def intersected_list(list1, list2):
    new_list = []
    for i in list1:
        if i in list2:
            new_list.append(i)
    return new_list

print(intersected_list([1, 2, 3], [2, 3, 4]))
```
