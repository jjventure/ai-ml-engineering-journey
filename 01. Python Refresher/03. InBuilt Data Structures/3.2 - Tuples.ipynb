### Introduction to Tuples

#### Explaination:

Tuples are ordered collection of items that are immutable. They are similar to lists, but their immutability makes them different 

```python
## creating a tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(lst)

tpl = tuple()
print(tpl)
```

```python
numbers = tuple([1, 2, 3, 4, 5, 6])
print(numbers)
```

```python
mixed_tuple = (1, "Hello", 3.14, True)
print(mixed_tuple)
print(type(mixed_tuple))
```

```python
#### Accessing tuple elements

numbers
```

```python
print(numbers[2])
print(numbers[-1])
```

```python
numbers[0:4]
```

```python
numbers[::]
```

```python
numbers[::-1]
```

```python
### Tuple operations

# Concatenation
concatenation_tuple = numbers + mixed_tuple
print(concatenation_tuple)
```

```python
mixed_tuple * 3
```

```python
numbers * 3
```

```python
### Immutable nature of tuples

## tuples are immutable - elements cant be changed once assigned

lst = [1, 2, 3, 4, 5, 6]
print(lst)

lst[1] = "Krish"    # lists are mutable
print(lst)
```

```python
numbers[1] = "Krish"   # immutability
```

```python
### Tuple Methods

numbers = (1, 2, 3, 4, 5, 6)

print(numbers.count(1))  # number of times 1 is present in the tuple: numbers

print(numbers.index(5))  # first occurence of the value 
```

```python
## packing and unpacking tuple

packed_tuple = 1, "Hello", 3.14   # by default, it becomes a tuple
print(packed_tuple)
```

```python
### unpacking a tuple

a, b, c = packed_tuple

print(a)
print(b)
print(c)
```

```python
### Unpacking with *

numbers = (1, 2, 3, 4, 5, 6)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

```python
### Example for nested tuples

# Example for nested lists 
lst = [[1, 2, 3, 4], [6, 7, 8, 9], [1, "Hello", 3.14, "c"]]
print(lst[0][:3])
```

```python
lst = [[1, 2, 3, 4], [6, 7, 8, 9], (1, "Hello", 3.14, "c")]
print(lst[2][:3])
```

```python
nested_tuple = (
    (1, 2, 3),
    ("a", "b", "c"),
    (True, False)
)

## accessing elements in nested tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end = " ")
    print()
```
