**Question 4:** Write a Python program to find the largest of three numbers.

```python
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))
num3 = int(input("Enter num3: "))

if (num1 >= num2) and (num1 >= num3):
    print(f"{num1} is the largest of three numbers.")
elif (num2 >= num1) and (num2 >= num3):
    print(f"{num2} is the largest of three numbers.")
else:
    print(f"{num3} is the largest of three numbers.")
```

**Question 5:** Write a Python program to calculate the factorial of a number.

```python
def factorial(n):
    if n == 0:
        return 1 
    else:
        return n * factorial(n-1)

num = int(input("Enter a number: "))
print(f"The factorial of {num} is {factorial(num)}")
```

**Question 8:** Write a Python program to convert Celsius to Fahrenheit.

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return f"{celsius}°C is equal to {fahrenheit}°F"

print(celsius_to_fahrenheit(10.0))
```

**Question 11:** Write a Python program to perform arithmetic operations: addition, subtraction, multiplication, and division.

```python
# Your code here
class Arithmetic_Operations:

    def add(num1, num2):
        result = num1 + num2
        return result 
    
    def subtract(num1, num2):
        result = num1 - num2 
        return result

    def multiplication(num1, num2):
        result = num1 * num2
        return result

    def division(num1, num2):
        result = num1 / num2
        return result

print(Arithmetic_Operations.add(1, 2))
print()
print(Arithmetic_Operations.subtract(2, 1))
print()
print(Arithmetic_Operations.multiplication(2, 2))
print()
print(Arithmetic_Operations.division(10, 2))    
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year!")
else:
    print("Not a Leap Year...")
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
string = input("Enter a string: ")

cleaned = string.replace(" ", "").lower()

if cleaned == cleaned[::-1]:
    print(f"The string '{string}' is palindrome!")
else:
    print(f"The string '{string}' is Not a palindrome...")
```
