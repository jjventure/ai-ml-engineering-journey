### Assignment 1: Reading a File

Write a function that reads the contents of a file named `sample.txt` and prints each line.

```python
def read_content(file_name):
    with open(file_name, 'r') as file:
        content = file.readlines()
        # print each line 
        for line in content:
            print(line)

read_content("sample.txt")
```

### Assignment 2: Writing to a File

Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

```python
def write_to_file(filename, lines):
    # Overwrites
    with open(filename, 'w') as file:
        for line in lines:
            file.write(line + "\n")
```

```python
write_to_file("output.txt", ["hello", "world"])
```

### Assignment 3: Copying a File

Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

```python
def copy_content(source_file, destination_file):
    with open(source_file, "r") as src:
        with open(destination_file, "w") as dest:
            dest.write(src.read())

copy_content("source.txt", "destination.txt")
```

### Assignment 5: Counting Words in a File

Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.

```python
def word_count(filename):
    with open(filename, 'r') as file:
        words = file.read().split()

        return len(words)

word_count("document.txt")
```

### Assignment 6: Finding and Replacing Text

Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
filename = "data.txt"

def replace_word(filename, given_word, new_word):
    with open(filename, "r") as file:
        content = file.read()
    content = content.replace(given_word, new_word)
    with open(filename, "w") as file:
        file.write(content)

replace_word(filename, "Python", "Java")
```

### Assignment 7: Reading a File in Reverse

Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

```python
def reverse_line_order(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
        for line in lines[::-1]:
            print(line.strip())

reverse_line_order("reverse.txt")
```

### Assignment 8: Counting Lines, Words, and Characters

Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
def count_lines_words_chars(filename):
    with open(filename, "r") as file:
        content = file.read()

        # chars 
        char_count = len(content) 

        # words 
        word_count = len(content.split())

        # lines 
        file.seek(0)
        lines = file.readlines()
        line_count = len(lines)
    
    return line_count, word_count, char_count


count_lines_words_chars("stats.txt")
```

```python
# interview approach 

def count_lines_words_characters(filename):
    with open(filename, "r") as file:
        content = file.read()

        line_count = len(content.splitlines())
        word_count = len(content.split())
        char_count = len(content)

    return line_count, word_count, char_count

print(count_lines_words_characters("stats.txt"))
```

### Assignment 9: Merging Multiple Files

Write a function that merges the contents of multiple files into a single file named `merged.txt`.

```python
def merge_multiple_files(files:list, output_file):
    with open(output_file, "w") as output_file:
        for file in files:
            with open(file, "r") as input_file:
                content = input_file.read()
                output_file.write(content + "\n")

merge_multiple_files(["file1.txt", "file2.txt", "file3.txt"], "merged.txt")
```

### Assignment 10: Splitting a Large File

Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

```python
# script for 1000 lines in large.txt file 
with open("large.txt", "w") as file:
    for i in range(1, 1001):
        file.write(f"This is line {i}.\n")
```

```python
def split_large_file(filename):
    """ splits large file to mini files each containing 100 lines each. """
    with open(filename, "r") as large_file:
        lines = large_file.readlines()
    for i in range(0, len(lines), 100):
        chunk = lines[i: i + 100]
        part_name = f"part{(i // 100) + 1}.txt"
        with open(part_name, "w") as part_file:
            part_file.writelines(chunk)


# function call 
split_large_file("large.txt")
```

### Assignment 13: CSV File Operations

Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

```python
# read csv file --> print(content) --> catch: list of dicts - [{}, {}, {}]

import csv 

def read_csv(csv_file):
    
    result = []

    with open(csv_file, "r") as file:
        reader = csv.DictReader(file)  # first row becomes keys and remaining values fills in correspondingly returning every row as a dict.

        for row in reader:
            result.append(row)

    return result

print(read_csv("data.csv"))
```

### Assignment 15: File Permission Handling

Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

```python
def read_protected_file(filename):
    try:
        with open(filename, "r") as file:
            content = file.read()
            print(content)

    except FileNotFoundError:
        print("File Not found.")

    except PermissionError as e:
        print(f"{e}: Sorry, you don't have access to this file!")

read_protected_file("protected.txt")
```
