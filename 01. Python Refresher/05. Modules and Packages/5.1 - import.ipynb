### Importing Modules in Python: Modules and Packages

In Python, modules and packages help organize and reuse code. Here's how to import them.

```python
import math

math.sqrt(16)
```

```python
from math import sqrt, pi

print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np 
```

```python
np.array([1, 2, 3, 4])
```

```python
from math import *  # * imports all functions in the module 
print(sqrt(81))
print(pi)
```

```python
from package.maths import addition

addition(5, 10)
```

```python
from package import maths

print(maths.addition(2, 3))
```
