# Pandas — DataFrame and Series

Pandas is a powerful Python library widely used for **data manipulation, data analysis, and data cleaning**.

It provides two primary data structures:

* **Series:** A one-dimensional, array-like object.
* **DataFrame:** A two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
## Series 
## A pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.

data = [1, 2, 3, 4, 5]
series = pd.Series(data)

print(f"Series\n{series}")
print(type(series))
```

```python
## Create a series from dictionary

data = {
    "a": 1,
    "b": 2,
    "c": 3
}

series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
indexes = ["a", "b", "c"]

pd.Series(data, index = indexes)
```

```python
## DataFrame 

## Create a df from a dict of list 

data = {
    "Name": ["Krish", "John", "Jack"],
    "Age": [25, 30, 45],
    "City": ["Bangalore", "New York", "Florida"]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Create a df from list of dictionaries

data = [
    {
        "Name": "Krish",
        "Age": 32,
        "City": "Bangalore"
    },
    {
        "Name": "John",
        "Age": 34,
        "City": "Bangalore"
    },
    {
        "Name": "Bappy",
        "Age": 32,
        "City": "Bangalore"
    },
    {
        "Name": "Jack",
        "Age": 32,
        "City": "Bangalore"
    }
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
