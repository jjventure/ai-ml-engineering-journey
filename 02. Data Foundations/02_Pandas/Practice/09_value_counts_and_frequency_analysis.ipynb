## 1. Understanding `value_counts()`

The Pandas `value_counts()` method returns the frequency of each unique value in a Series.

In simple terms, it tells us **how many times each distinct value occurs**.

### Example

```python
import pandas as pd

department_data = {
    "Employee": ["Alice", "Bob", "Charlie", "Diana", "Ethan",
                 "Fiona", "George", "Helen", "Ian", "Jack"],
    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "IT", "Finance", "HR", "IT", "Finance"]
}

department_df = pd.DataFrame(department_data)
```

Count the occurrences of each department:

```python
department_counts = department_df["Department"].value_counts()

department_counts
```

### Expected Output

```text
Department
IT         4
HR         3
Finance    3
Name: count, dtype: int64
```

### Understanding the Output

- IT appears 4 times.
- HR appears 3 times.
- Finance appears 3 times.

By default, `value_counts()` sorts the results by frequency in descending order.

**Important:** `value_counts()` returns a Pandas Series, not a DataFrame.

## 2. Key Syntax

```python
series.value_counts()
```

You call this method directly on a Pandas Series, usually a DataFrame column.

## 2. Important Parameters of `value_counts()`

The `value_counts()` method provides several parameters to control how frequency counts are calculated and displayed.

We'll cover them individually.

### 1. `normalize`

By default, `value_counts()` returns absolute frequencies (actual counts).

Setting `normalize=True` returns relative frequencies (proportions) instead.

**Example:**

```python
department_df["Department"].value_counts(normalize=True)
```

Expected output:

```text
Department
IT         0.4
HR         0.3
Finance    0.3
Name: proportion, dtype: float64
```

Each value represents the proportion of the total observations belonging to that category.

To display percentages, multiply the result by 100:

```python
department_df["Department"].value_counts(normalize=True) * 100
```

### 2. `sort`

By default, `sort=True`.

This sorts the results by frequency in descending order.

Setting `sort=False` preserves the order in which unique values first appear in the original Series.

```python
department_df["Department"].value_counts(sort=False)
```

### 3. `ascending`

By default, `ascending=False`.

This displays the least frequent values first when set to `True`.

```python
department_df["Department"].value_counts(ascending=True)
```

**Note:** `ascending` controls the direction of frequency sorting when `sort=True`.

### 4. `dropna`

By default, `dropna=True`.

Missing values are excluded from the frequency counts.

Setting `dropna=False` includes missing values as a separate entry.

```python
department_df["Department"].value_counts(dropna=False)
```

### Parameter Summary

| Parameter | Default | Purpose |
|---|---|---|
| `normalize` | `False` | Returns proportions instead of counts |
| `sort` | `True` | Sorts by frequency |
| `ascending` | `False` | Controls frequency sorting direction |
| `dropna` | `True` | Controls whether missing values are included |

````python
department_df
````

````python
# proportion
department_df["Department"].value_counts(normalize=True)
````

````python
# Display percentages
department_df["Department"].value_counts(normalize=True) * 100
````

````python
### sort 
# By default: sort = True
# This sorts the results by frequency in descending order.

department_df["Department"].value_counts(sort=False)
````

````python
### ascending 
# By default: ascending = False
# This displays the least frequent values first when set to True.

department_df["Department"].value_counts(ascending=True)
````

````python
### dropna
# By default, dropna = True 
# Missing values are excluded from the frequency counts.
# Setting dropna=False includes missing values as a seperate entry. 

department_df["Department"].value_counts(dropna=False)
````
