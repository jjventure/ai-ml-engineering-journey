## Lesson 1 — The `.query()` Method

So far, we've filtered DataFrames using Boolean indexing:

```python
employee_df[employee_df["Age"] > 25]
```

Pandas provides another approach: `.query()`.

### Syntax

```python
df.query("condition")
```

### Example

```python
employee_df.query("Age > 25")
```

This returns employees whose age is greater than 25.

### Important Operators

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| `>`      | Greater than                        |
| `<`      | Less than                           |
| `>=`     | Greater than or equal to            |
| `<=`     | Less than or equal to               |
| `==`     | Equal to                            |
| `!=`     | Not equal to                        |
| `and`    | Both conditions must be true        |
| `or`     | At least one condition must be true |
| `not`    | Negation                            |

# Task 3 — Using Variables inside `.query()`

### Objective

Learn how to pass external Python variables into Pandas `.query()` expressions using the `@` operator.

## 1. Why do we need this?

Previously, we used fixed values directly inside `.query()`:

```python
employee_filter_df.query("Age > 25")
```

But what if the filtering value changes?

Instead of modifying the query every time, we can use Python variables.

## 2. Syntax

```python
variable_name = value

df.query("column_name > @variable_name")
```

**Important:** The `@` symbol tells Pandas to retrieve the value from an external Python variable.

## 3. Example

```python
minimum_age = 28

employee_filter_df.query("Age >= @minimum_age")
```

This retrieves employees whose age is greater than or equal to the value stored in `minimum_age`.

# Task 4 — Filtering with `.isin()`

### Objective

Learn to filter DataFrame rows based on whether column values belong to a specified collection.

## 1. What is `.isin()`?

Suppose you want to retrieve employees belonging to either the IT or Finance department.

You could write:

```python
employee_filter_df.query(
    "Department == 'IT' or Department == 'Finance'"
)
```

But Pandas provides a simpler approach: `.isin()`.

## 2. Syntax

```python
df[df["column"].isin(values)]
```

Here:

* `df["column"]` selects the column.
* `.isin(values)` checks whether each value exists in the supplied collection.
* The resulting Boolean Series filters the DataFrame.

## 3. Example

```python
selected_departments = ["IT", "Finance"]

employee_filter_df[
    employee_filter_df["Department"].isin(selected_departments)
]
```

This retrieves employees belonging to either IT or Finance.

## 4. Understanding the Output

Consider:

```python
employee_filter_df["Department"].isin(["IT", "Finance"])
```

This produces a Boolean Series:

| Index | Department | Result |
| ----- | ---------- | ------ |
| 0     | IT         | True   |
| 1     | HR         | False  |
| 2     | IT         | True   |
| 3     | Finance    | True   |
| 4     | HR         | False  |
| 5     | Finance    | True   |

Only rows returning `True` are retained.
