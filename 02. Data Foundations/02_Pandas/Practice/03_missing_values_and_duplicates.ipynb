### Exercise 11 — Detecting Missing Values

**Objective:** Identify missing values in a Pandas DataFrame.

Create a new DataFrame named `student_df` using the following dataset:

| Name    | Age | Marks |
| ------- | --: | ----: |
| Alice   |  20 |    85 |
| Bob     |  21 |   NaN |
| Charlie | NaN |    78 |
| Diana   |  22 |    92 |
| Ethan   |  20 |   NaN |

Here, `NaN` represents a missing value.

**Tasks:**

1. Create the DataFrame using a Python dictionary.
2. Display the complete DataFrame.
3. Identify which values are missing.
4. Count the missing values in each column.
5. Calculate the total number of missing values across the entire DataFrame.

**Requirements:**

* Use Pandas to perform the analysis.
* Do not manually count the missing values.
* Keep the original data unchanged.

**Learning goal:** Understand how to identify and quantify missing data before deciding how to handle it.


```python
# 1. Create the df using python dict 
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan"],
    "Age": [20, 21, None, 22, 20],
    "Marks": [85, None, 78, 92, None]
}
```

```python
# 2. Display the complete dataframe 
student_df = pd.DataFrame(data)
student_df
```

```python
# 3. Identify which values are missing
student_df.isnull()
```

```python
# 4. Count the missing values in each column
student_df.isnull().sum()
```

```python
# 5. Calculate the total number of missing values in the entire dataframe 
student_df.isnull().sum().sum()
```

### Exercise 12 — Handling Missing Values

**Objective:** Practise removing rows containing missing values and filling missing values with appropriate replacements.

Use the existing `student_df` DataFrame.

**Tasks:**

1. Create a new DataFrame named `students_complete` containing only rows without missing values. Do not modify `student_df`.
2. Display `students_complete`.
3. Create another DataFrame named `students_filled` by replacing missing values in the `Age` column with the average age.
4. Replace missing values in the `Marks` column with the average marks.
5. Display `students_filled` and verify that neither column contains missing values.

**Requirements:**

* Use Pandas methods rather than manually editing individual rows.
* Use `dropna()` to remove rows containing missing values.
* Use `fillna()` to replace missing values.
* Calculate the averages from the available values in the original DataFrame.
* Keep `student_df` unchanged.

**Learning goal:** Understand the difference between removing incomplete observations and imputing missing values.


```python
student_df
```

```python
# dropna() returns a new DataFrame by default
students_complete = student_df.dropna()
students_complete
```

```python
students_filled = student_df.copy()
mean_age = students_filled['Age'].mean()
students_filled["Age"] = students_filled["Age"].fillna(mean_age)
students_filled
```

```python
mean_marks = students_filled["Marks"].mean()
students_filled["Marks"] = students_filled["Marks"].fillna(mean_marks)
students_filled
```

### Exercise 13 — Detecting and Removing Duplicates

**Objective:** Identify duplicate records and remove them from a DataFrame.

Create a new DataFrame named `sales_df` using the following dataset:

| Name    | Product | Amount |
| ------- | ------- | -----: |
| Alice   | Laptop  |  50000 |
| Bob     | Phone   |  20000 |
| Alice   | Laptop  |  50000 |
| Charlie | Tablet  |  15000 |
| Bob     | Phone   |  20000 |
| Diana   | Monitor |  12000 |

**Tasks:**

1. Create the DataFrame using a Python dictionary.
2. Display the complete DataFrame.
3. Identify duplicate rows.
4. Count the number of duplicate rows.
5. Create a new DataFrame named `sales_unique` with duplicate rows removed.
6. Display `sales_unique` and verify that no duplicate rows remain.

**Requirements:**

* Use Pandas methods to detect and remove duplicates.
* Do not manually identify or delete individual rows.
* Keep the original `sales_df` unchanged.

**Learning goal:** Understand how to detect repeated records and remove duplicates while preserving the original dataset.


```python
import pandas as pd 

# 1. Create the dataframe using python dictionary
sales_data = {
    "Name": ["Alice", "Bob", "Alice", "Charlie", "Bob", "Diana"],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Monitor"],
    "Amount": [50000, 20000, 50000, 15000, 20000, 12000]
}

sales_df = pd.DataFrame(sales_data)

# 2. Display the complete dataframe
sales_df
```

```python
# 3. Identify duplicate rows 
sales_df.duplicated()
```

```python
# 4. Count the number of duplicate rows
sales_df.duplicated().sum()
```

```python
# 5. Create a new DataFrame named sales_unique with duplicate rows removed
sales_unique = sales_df.copy()

sales_unique = sales_unique.drop_duplicates()
```

```python
# 6. Display `sales_unique` and verify that no duplicate rows remain
print(sales_unique)
print(f"\nNumber of duplicated rows: {sales_unique.duplicated().sum().sum()}")
```
