# 1. DataFrame Concatenation — `pd.concat()`

## What is Concatenation?

Concatenation is the process of combining two or more DataFrames along a particular axis.

In Pandas, we use `pd.concat()` to combine DataFrames.

This is useful when data is stored across multiple DataFrames but needs to be brought together for analysis.

### Basic Syntax

```python
pd.concat(
    [df1, df2],
    axis=0,
    ignore_index=False
)
```

**Parameters:**

- `objs`: A list or collection of DataFrames to concatenate.
- `axis=0`: Combines DataFrames vertically (row-wise). This is the default.
- `axis=1`: Combines DataFrames horizontally (column-wise).
- `ignore_index=False`: Preserves the original index labels.
- `ignore_index=True`: Creates a new continuous integer index.

We'll explore these parameters individually through practical examples.

## Two Main Types of Concatenation

1. **Vertical Concatenation (`axis=0`)** — Appends rows from one DataFrame below another.
2. **Horizontal Concatenation (`axis=1`)** — Places columns from one DataFrame beside another.

**Important:** `concat()` combines DataFrames based on their axis and labels. It does not match records using common key columns like `merge()` does.

````python
import pandas as pd

employee_data_1 = {
    "Employee_ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"],
    "Department": ["IT", "HR", "Finance"]
}

employee_data_2 = {
    "Employee_ID": [104,105, 106],
    "Name": ["Diana", "Ethan", "Fiona"],
    "Department": ["IT", "Finance", "HR"]
}

employee_df_1 = pd.DataFrame(employee_data_1)
employee_df_2 = pd.DataFrame(employee_data_2)

display(employee_df_1)
display(employee_df_2)
````

## 1.1 Vertical Concatenation — `axis=0`

Vertical concatenation combines DataFrames by appending rows from one DataFrame below another.

By default, `pd.concat()` uses `axis=0`.

### Example

```python
combined_employee_df = pd.concat(
    [employee_df_1, employee_df_2],
    axis=0
)

display(combined_employee_df)
```

### What happens?

- Rows from `employee_df_2` are appended below the rows of `employee_df_1`.
- Columns are aligned using their column names.
- The original index labels are preserved by default.
- The original DataFrames remain unchanged.

**Important:** When concatenating vertically, Pandas aligns columns by their column labels, not simply by their positions.

````python
combined_employee_df = pd.concat(
    [employee_df_1, employee_df_2],
    axis=0
)

display(combined_employee_df)
````

````python
# Using ignore_index=True:

combined_employee_df_reset = pd.concat(
    [employee_df_1, employee_df_2],
    axis=0,
    ignore_index=True
)

display(combined_employee_df_reset)
````

## 1.2 Horizontal Concatenation — `axis=1`

Horizontal concatenation combines DataFrames side by side by adding columns.

We use `axis=1` in `pd.concat()` to concatenate DataFrames horizontally.

### Example

First, create two DataFrames:

```python
employee_details = {
    "Employee_ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"]
}

employee_salary = {
    "Salary": [50000, 60000, 65000],
    "Experience": [2, 4, 5]
}

employee_details_df = pd.DataFrame(employee_details)
employee_salary_df = pd.DataFrame(employee_salary)
```

Display both DataFrames:

```python
display(employee_details_df)
display(employee_salary_df)
```

### Horizontal Concatenation

```python
employee_full_details_df = pd.concat(
    [employee_details_df, employee_salary_df],
    axis=1
)

display(employee_full_details_df)
```

### What happens?

- Columns from both DataFrames are placed side by side.
- Pandas aligns rows using their index labels.
- The original index labels are preserved by default.
- The original DataFrames remain unchanged.

### Important: Index Alignment

Unlike vertical concatenation, horizontal concatenation aligns DataFrames using their row indexes.

For example, if one DataFrame has indexes `[0, 1, 2]` and another has indexes `[1, 2, 3]`, Pandas aligns matching index labels.

Any index without a corresponding match receives missing values (`NaN`).

**Remember:**
- `axis=0` → Combines rows vertically.
- `axis=1` → Combines columns horizontally.

````python
employee_details = {
    "Employee_ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"]
}

employee_salary = {
    "Salary": [50000, 60000, 65000],
    "Experience": [2, 4, 5]
}

employee_details_df = pd.DataFrame(employee_details)
employee_salary_df = pd.DataFrame(employee_salary)
````

````python
employee_salary_df
````

````python
# Horizontal Concatenation (Column-wise)

employee_full_details_df = pd.concat(
    [employee_details_df, employee_salary_df],
    axis=1
)

display(employee_full_details_df)
````

````python
# Horizontal concatenation with mismatched indexes

employee_details = {
    "Employee_ID": [101, 102, 103]
}

employee_salary = {
    "Salary": [60000, 65000, 70000]
}

employee_details_df = pd.DataFrame(
    employee_details,
    index=[0, 1, 2]
)

employee_salary_df = pd.DataFrame(
    employee_salary,
    index=[1, 2, 3]
)
````

````python
employee_mismatched_df = pd.concat(
    [employee_details_df, employee_salary_df],
    axis=1
)

display(employee_mismatched_df)
````

````python
# Let's compare join='outer' and join='inner'.

# 1. join='outer' — Keep all index labels
# This is the default behaviour 

employee_outer_df = pd.concat(
    [employee_details_df, employee_salary_df],
    axis=1,
    join="outer"
)

display(employee_outer_df)

# It retains every index found in either DataFrame. Missing matches are filled with NaN.
````

````python
# 2. join='inner' — Keep only common index labels

employee_inner_df = pd.concat(
    [employee_details_df, employee_salary_df],
    axis=1,
    join="inner"
)

display(employee_inner_df)

# It retains only index labels present in both DataFrames.
````

### Vertical Concatenation with Mismatched Columns

When concatenating DataFrames vertically (`axis=0`), Pandas matches columns using their column names.

If one DataFrame does not contain a column present in another DataFrame, Pandas fills the missing values with `NaN`.

By default, `pd.concat()` uses `join='outer'`, which retains all columns from both DataFrames.

**Key points:**
- Columns are aligned by column names.
- All columns from both DataFrames are retained by default.
- Missing column values are filled with `NaN`.
- `join='inner'` retains only columns common to both DataFrames.

````python
department_it = pd.DataFrame({
    "Employee_ID": [101, 102],
    "Name": ["Alice", "Bob"],
    "Department": ["IT", "IT"]
})

department_hr = pd.DataFrame({
    "Employee_ID": [103, 104],
    "Name": ["Charlie", "Diana"],
    "Salary": [50000, 55000]
})
````

````python
display(department_it)
````

````python
display(department_hr)
````

````python
# default --> join="outer"
employee_outer_columns_df = pd.concat(
    [department_it, department_hr],
    axis=0,
    ignore_index=True
)

display(employee_outer_columns_df)
````

````python
# join="inner"
employee_inner_columns_df = pd.concat(
    [department_it, department_hr],
    axis=0,
    join="inner",
    ignore_index=True
)

display(employee_inner_columns_df)
````

### Using `keys` in `pd.concat()`

The `keys` parameter adds labels to identify the source of each DataFrame during concatenation.

**Syntax:**
```python
pd.concat([df1, df2], keys=["source1", "source2"])
```

**Key points:**
- Creates a hierarchical (MultiIndex) index.
- The outer index identifies the source DataFrame.
- The inner index preserves the original row index.
- Useful for tracking the origin of combined data.

````python
employee_keyed_df = pd.concat(
    [department_it, department_hr],
    keys=["IT_Records", "HR_Records"]
)
display(employee_keyed_df)
````

## 2. Merging DataFrames — `pd.merge()`

`pd.merge()` combines two DataFrames by matching records using one or more common columns, known as **keys**.

It is similar to SQL JOIN operations in relational databases.

**Syntax:**

```python
pd.merge(
    left,
    right,
    how="inner",
    on=None
)
```

**Important parameters:**

| Parameter | Purpose |
|---|---|
| `left` | First DataFrame |
| `right` | Second DataFrame |
| `how` | Specifies the type of join |
| `on` | Column name(s) used to match records |

### How is `merge()` different from `concat()`?

- `concat()` combines DataFrames along rows or columns.
- `merge()` matches related records using common key columns.

**Example:** Suppose one DataFrame contains employee details and another contains employee salaries. Both contain `Employee_ID`.

We can merge them using `Employee_ID` to create a DataFrame containing both employee details and salaries.

````python
employee_details_data = {
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Department": ["IT", "HR", "Finance", "IT"]
}

employee_salary_data = {
    "Employee_ID": [101, 102, 103, 105],
    "Salary": [60000, 50000, 65000, 70000],
    "Experience": [3, 2, 5, 6]
}

employee_details_df = pd.DataFrame(employee_details_data)
employee_salary_df = pd.DataFrame(employee_salary_data)

print("employee_details_df: ")
display(employee_details_df)

print("employee_salary_df: ")
display(employee_salary_df)
````

### 2.1 Inner Join — `how="inner"`

An **Inner Join** returns only the records that have matching key values in both DataFrames.

In Pandas, `how="inner"` is the default behaviour of `pd.merge()`.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    on="common_column",
    how="inner"
)
```

**Key points:**
- Matches records using the specified key column.
- Retains only matching records from both DataFrames.
- Excludes records without a match.
- Combines columns from both DataFrames.

````python
### Perform an inner join 

employee_inner_merge_df = pd.merge(
    employee_details_df, 
    employee_salary_df,
    on="Employee_ID",
    how="inner"
)

display(employee_inner_merge_df)
````

### 2.2 Left Join — `how="left"`

A **Left Join** retains all records from the left DataFrame and adds matching records from the right DataFrame.

If a record in the left DataFrame has no matching key in the right DataFrame, the corresponding right-side columns contain `NaN`.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    on="common_column",
    how="left"
)
```

**Key points:**
- Retains every record from the left DataFrame.
- Adds matching records from the right DataFrame.
- Unmatched right-side values become `NaN`.
- Records existing only in the right DataFrame are excluded.

````python
### Perform a left join 

employee_left_merge_df = pd.merge(
    employee_details_df,
    employee_salary_df,
    on="Employee_ID",
    how="left"
)

display(employee_left_merge_df)
````

### 2.3 Right Join — `how="right"`

A **Right Join** retains all records from the right DataFrame and adds matching records from the left DataFrame.

If a record in the right DataFrame has no matching key in the left DataFrame, the corresponding left-side columns contain `NaN`.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    on="common_column",
    how="right"
)
```

**Key points:**
- Retains every record from the right DataFrame.
- Adds matching records from the left DataFrame.
- Unmatched left-side values become `NaN`.
- Records existing only in the left DataFrame are excluded.

````python
employee_right_merge_df = pd.merge(
    employee_details_df,
    employee_salary_df,
    on="Employee_ID",
    how="right"
)

display(employee_right_merge_df)
````

### 2.4 Full Outer Join — `how="outer"`

A **Full Outer Join** retains all records from both DataFrames.

- Matching records are combined.
- Records without a match are also retained.
- Missing values are filled with `NaN`.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    on="common_column",
    how="outer"
)
```

**Key points:**
- Retains every record from both DataFrames.
- Combines matching records.
- Unmatched values become `NaN`.
- Neither DataFrame is prioritised.

````python
employee_outer_merge_df = pd.merge(
    employee_details_df,
    employee_salary_df,
    on="Employee_ID",
    how="outer"
)

display(employee_outer_merge_df)
````

### 2.5 Merging with Different Key Column Names

When the matching key columns have different names in the two DataFrames, use `left_on` and `right_on`.

**Parameters:**

- `left_on`: Column name from the left DataFrame.
- `right_on`: Column name from the right DataFrame.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    left_on="left_column",
    right_on="right_column",
    how="inner"
)
```

**Example:**

Suppose:
- The left DataFrame contains `Employee_ID`.
- The right DataFrame contains `ID`.

Both columns represent the same employee identifier.

We can merge them using:

```python
pd.merge(
    employee_details_df,
    employee_salary_df,
    left_on="Employee_ID",
    right_on="ID"
)
```

````python
employee_info = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "Diana"]
})

employee_performance = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Performance_Score": [90, 85, 95, 80]
})

print("employee_info: ")
display(employee_info)

print("employee_performance: ")
display(employee_performance)
````

````python
### Perform the merge

employee_performance_merge_df = pd.merge(
    employee_info,
    employee_performance,
    left_on="Employee_ID",
    right_on="ID",
    how="inner"
)

display(employee_performance_merge_df)
````

### 2.6 Handling Duplicate Column Names — `suffixes`

When merging two DataFrames, both may contain columns with identical names other than the merge key.

Pandas automatically adds suffixes to distinguish these columns.

By default, the suffixes are `('_x', '_y')`.

**Example:**

Suppose both DataFrames contain a column named `Department`.

After merging, Pandas produces:
- `Department_x` — from the left DataFrame.
- `Department_y` — from the right DataFrame.

We can customize these suffixes using the `suffixes` parameter.

**Syntax:**

```python
pd.merge(
    left_df,
    right_df,
    on="Employee_ID",
    suffixes=("_left", "_right")
)
```

This produces:
- `Department_left`
- `Department_right`

````python
employee_department_1 = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"],
    "Department": ["IT", "HR", "Finance"]
})

employee_department_2 = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Salary": [60000, 50000, 65000],
    "Department": ["Technology", "Human Resources", "Finance"]
})

display(employee_department_1)
display(employee_department_2)
````

````python
# Merge using default suffixes

employee_suffix_default_df = pd.merge(
    employee_department_1,
    employee_department_2,
    on="Employee_ID"
)

display(employee_suffix_default_df)
````

````python
# Customize suffixes

employee_suffix_custom_df = pd.merge(
    employee_department_1,
    employee_department_2,
    on="Employee_ID",
    suffixes=("_original", "_updated")
)

display(employee_suffix_custom_df)
````

## 3. DataFrame.join()

### 3.1 Understanding `DataFrame.join()`

Pandas provides the `.join()` method to combine two DataFrames, primarily using their indexes.

It is particularly useful when two DataFrames share the same index, such as an employee ID, customer ID, or product ID.

**Difference between concat(), merge(), and join():**

| Operation | Purpose |
|---|---|
| `pd.concat()` | Combines DataFrames along rows or columns. |
| `pd.merge()` | Combines DataFrames using matching key columns. |
| `.join()` | Combines DataFrames primarily using their indexes. |

### Syntax

```python
left_df.join(
    other,
    how="left",
    lsuffix="",
    rsuffix=""
)
```

### Important Parameters

- `other`: The DataFrame to join with.
- `how`: Specifies the type of join. Supports `left`, `right`, `inner`, and `outer`.
- `lsuffix`: Suffix added to overlapping column names from the left DataFrame.
- `rsuffix`: Suffix added to overlapping column names from the right DataFrame.

**Note:** By default, `.join()` performs a left join using the indexes of both DataFrames.

````python
employee_info = pd.DataFrame(
    {
        "Name": ["Alice", "Bob", "Charlie", "Diana"]
    },
    index=[101, 102, 103, 104]
)

employee_salary = pd.DataFrame(
    {
        "Salary": [60000, 50000, 65000, 70000]
    },
    index=[101, 102, 103, 105]
)

print("employee_info: ")
display(employee_info)

print("employee_salary: ")
display(employee_salary)
````

### 3.2 Performing an Inner Join

An inner join retains only the records whose index values are present in both DataFrames.

Unmatched index values are excluded from the result.

````python
employee_inner_join_df = employee_info.join(
    employee_salary,
    how="inner"
)

display(employee_inner_join_df)
````

### 3.3 Types of Joins

Similar to `pd.merge()`, the `.join()` method supports four common join types.

| Join Type | Description |
|---|---|
| `left` | Retains all indexes from the left DataFrame. |
| `right` | Retains all indexes from the right DataFrame. |
| `inner` | Retains only indexes present in both DataFrames. |
| `outer` | Retains all indexes from both DataFrames. |

**Important:** Unmatched indexes produce `NaN` values in the columns where corresponding data is unavailable.

By default, `.join()` uses `how="left"`.

````python
# left join

employee_left_join_df = employee_info.join(
    employee_salary,
    how="left"
)

display(employee_left_join_df)
````

````python
# right join

employee_right_join_df = employee_info.join(
    employee_salary,
    how="right"
)

display(employee_right_join_df)
````

````python
# outer join 

employee_outer_join_df = employee_info.join(
    employee_salary,
    how="outer"
)

display(employee_outer_join_df)
````

### 3.4 Handling Overlapping Column Names

When joining two DataFrames, both may contain columns with identical names.

By default, `.join()` raises a `ValueError` if overlapping column names exist.

We can resolve this using:

- `lsuffix`: Adds a suffix to overlapping columns from the left DataFrame.
- `rsuffix`: Adds a suffix to overlapping columns from the right DataFrame.

**Syntax:**

```python
left_df.join(
    right_df,
    lsuffix="_left",
    rsuffix="_right"
)
```

````python
employee_details = pd.DataFrame(
    {
        "Name": ["Alice", "Bob"],
        "Department": ["IT", "HR"]
    },
    index=[101, 102]
)

employee_updates = pd.DataFrame(
    {
        "Department": ["Technology", "Human Resources"],
        "Salary": [60000, 50000]
    },
    index=[101, 102]
)

employee_details.join(
    employee_updates,
    lsuffix="_original",
    rsuffix="_updated"
)
````
