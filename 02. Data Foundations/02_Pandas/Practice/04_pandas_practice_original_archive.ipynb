### Exercise 13 — Detecting and Removing Duplicates

**Objective:** Identify duplicate records and remove them from a DataFrame.

Create a new DataFrame named `sales_df` using the following dataset:

| Name    | Product | Amount |
| ------- | ------- | -----: |
| Alice   | Laptop  |  50000 |
| Bob     | Phone   |  20000 |
| Alice   | Laptop  |  50000 |
| Charlie | Tablet  |  15000 |
| Bob     | Phone   |  20000 |
| Diana   | Monitor |  12000 |

**Tasks:**

1. Create the DataFrame using a Python dictionary.
2. Display the complete DataFrame.
3. Identify duplicate rows.
4. Count the number of duplicate rows.
5. Create a new DataFrame named `sales_unique` with duplicate rows removed.
6. Display `sales_unique` and verify that no duplicate rows remain.

**Requirements:**

* Use Pandas methods to detect and remove duplicates.
* Do not manually identify or delete individual rows.
* Keep the original `sales_df` unchanged.

**Learning goal:** Understand how to detect repeated records and remove duplicates while preserving the original dataset.


````python
import pandas as pd 

# 1. Create the dataframe using python dictionary
sales_data = {
    "Name": ["Alice", "Bob", "Alice", "Charlie", "Bob", "Diana"],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Monitor"],
    "Amount": [50000, 20000, 50000, 15000, 20000, 12000]
}

sales_df = pd.DataFrame(sales_data)

# 2. Display the complete dataframe
sales_df
````

````python
# 3. Identify duplicate rows 
sales_df.duplicated()
````

````python
# 4. Count the number of duplicate rows
sales_df.duplicated().sum()
````

````python
# 5. Create a new DataFrame named sales_unique with duplicate rows removed
sales_unique = sales_df.copy()

sales_unique = sales_unique.drop_duplicates()
````

````python
# 6. Display `sales_unique` and verify that no duplicate rows remain
print(sales_unique)
print(f"\nNumber of duplicated rows: {sales_unique.duplicated().sum().sum()}")
````

# Task 4 — Filtering with `.isin()`

### Objective

Learn to filter DataFrame rows based on whether column values belong to a specified collection.

## 1. What is `.isin()`?

Suppose you want to retrieve employees belonging to either the IT or Finance department.

You could write:

```python
employee_filter_df.query(
    "Department == 'IT' or Department == 'Finance'"
)
```

But Pandas provides a simpler approach: `.isin()`.

## 2. Syntax

```python
df[df["column"].isin(values)]
```

Here:

* `df["column"]` selects the column.
* `.isin(values)` checks whether each value exists in the supplied collection.
* The resulting Boolean Series filters the DataFrame.

## 3. Example

```python
selected_departments = ["IT", "Finance"]

employee_filter_df[
    employee_filter_df["Department"].isin(selected_departments)
]
```

This retrieves employees belonging to either IT or Finance.

## 4. Understanding the Output

Consider:

```python
employee_filter_df["Department"].isin(["IT", "Finance"])
```

This produces a Boolean Series:

| Index | Department | Result |
| ----- | ---------- | ------ |
| 0     | IT         | True   |
| 1     | HR         | False  |
| 2     | IT         | True   |
| 3     | Finance    | True   |
| 4     | HR         | False  |
| 5     | Finance    | True   |

Only rows returning `True` are retained.
