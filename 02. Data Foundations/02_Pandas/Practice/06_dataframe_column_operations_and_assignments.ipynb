### Task 11 — Employee Data Cleaning Pipeline

**Objective:** Apply multiple Pandas operations to clean and transform an employee dataset.

Use the existing `employee_dirty_df` DataFrame.

**Tasks:**

1. Create an independent copy named `employee_final_df`.
2. Rename the `Name` column to `Employee_Name`.
3. Remove the `Salary` column.
4. Convert the `Age` column into an integer.
5. Use `.assign()` to create a new column called `Age_Group`:

   * Age below 30 → `"Young"`
   * Age 30 or above → `"Senior"`
6. Display the final DataFrame.
7. Display its data types.
8. Display the original `employee_dirty_df` to verify that it remains unchanged.

**Requirements:**

* Use `.copy()`.
* Use `.rename()`.

* Use `.astype()`.

* Do not modify the original DataFrame.

**Learning goal:** Combine DataFrame copying, column renaming, deletion, type conversion and conditional column creation into one workflow.


````python
employee_dirty_df
````

````python
import numpy as np

employee_final_df = employee_dirty_df.copy()

employee_final_df = employee_final_df.rename(
    columns={
        "Name": "Employee_Name"
    },
    errors="raise"
)

employee_final_df = employee_final_df.drop(
    columns = ["Salary"]
)

employee_final_df["Age"] = employee_final_df["Age"].astype(int)

employee_final_df = employee_final_df.assign(
    Age_Group = lambda df: np.where(
        df["Age"] < 30,
        "Young",
        "Senior"
    )
)

print(f"Final DataFrame:\n{employee_final_df}\n\n")
print(f"Data types:\n{employee_final_df.dtypes}\n\n")
print(f"Employee dirty DataFrame:\n{employee_dirty_df}")
````

````python
employee_data_2 = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Kiran", "Meera"],
    "Department": ["IT", "HR", "Finance", "IT", "HR", "Finance"],
    "Age": ["24", "32", "29", "27", "35", "26"],
    "Monthly_Salary": ["40000", "55000", "60000", "48000", "52000", "65000"],
    "Experience": [1, 8, 5, 3, 10, 4]
}

employee_dirty_df_2 = pd.DataFrame(employee_data_2)
````

````python
employee_dirty_df_2
````

````python
employee_dirty_df_2.dtypes
````

````python
import numpy as np

# copy
employee_final_df_2 = employee_dirty_df_2.copy()

# rename 
employee_final_df_2 = employee_final_df_2.rename(
    columns={
        "Name": "Employee_Name",
        "Monthly_Salary": "Salary"
    }
)

# drop 
employee_final_df_2 = employee_final_df_2.drop(
    columns=["Experience"]
)

# Convert data types
employee_final_df_2["Age"] = employee_final_df_2["Age"].astype(int)
employee_final_df_2["Salary"] = employee_final_df_2["Salary"].astype(float)

# Calculate Annual Salary
employee_final_df_2 = employee_final_df_2.assign(
    Annual_Salary = lambda df: df["Salary"] * 12 
)

# Calculate Bonus 
employee_final_df_2 = employee_final_df_2.assign(
    Bonus = lambda df: np.where(
        df["Annual_Salary"] >= 600000,
        df["Annual_Salary"] * 0.15,
        df["Annual_Salary"] * 0.05
    )
)

# Calculate Total Compensation
employee_final_df_2 = employee_final_df_2.assign(
    Total_Compensation = lambda df: df["Annual_Salary"] + df["Bonus"]
)

# Create Age Groups
employee_final_df_2 = employee_final_df_2.assign(
    Age_Group = lambda df: np.where(
        df["Age"] < 30,
        "Young",
        "Senior"
    )
)

# Remove Rows
employee_final_df_2 = employee_final_df_2[
    employee_final_df_2["Age"] >= 25
]

# reset index - so it starts running from 0.
employee_final_df_2 = employee_final_df_2.reset_index(drop=True)

# final df 
print(f"Final DataFrame:\n{employee_final_df_2}\n\n")

# final df's dtypes
print(f"Final DataFrame's Data Types:\n{employee_final_df_2.dtypes}\n\n")

# original employee_dirty_df_2
print(f"Original Employee Dirty DataFrame 2:\n{employee_dirty_df_2}\n\n")

# original DataFrame's Data Types
print(f"Original DataFrame's Data Types:\n{employee_dirty_df_2.dtypes}")
````
