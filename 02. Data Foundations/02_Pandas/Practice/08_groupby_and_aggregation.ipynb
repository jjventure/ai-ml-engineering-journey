## 20.1 — Understanding GroupBy

The `groupby()` operation divides a DataFrame into groups based on one or more columns.

It follows the **Split–Apply–Combine** approach:

1. **Split:** Divide the data into groups.
2. **Apply:** Perform an operation on each group.
3. **Combine:** Combine the results into a final output.

**Example:** Group employees by department and calculate the average salary of each department.


````python
employee_df
````

````python
employee_df.groupby("Department")
````

````python
employee_df.groupby("Department").groups
````

````python
# avg salary of each department 
department_avg_salary = employee_df.groupby("Department")["Salary"].mean()
department_avg_salary
````

````python
# total salary of each department
employee_df.groupby("Department")["Salary"].sum()
````

````python
# number of non-null salary entries 
employee_df.groupby("Department")["Salary"].count()
````

````python
# max salary of each department
employee_df.groupby("Department")["Salary"].max()
````

````python
# min salary of each department 
employee_df.groupby("Department")["Salary"].min()
````

````python
# basic syntax
employee_df.groupby("Department")["Salary"].agg(["mean", "sum", "min", "max"])
````

````python
department_salary_summary = employee_df.groupby("Department")["Salary"].agg(
    ["mean", "sum", "min", "max"]
)

department_salary_summary
````

````python
employee_df
````

One more useful feature: Different aggregations for different columns
So far, we've applied multiple aggregations to a single column (Salary).

But Pandas also allows us to aggregate multiple columns differently.

For example:

````python
employee_df.groupby("Department").agg(
    {
        "Salary": ["mean", "max"],
        "Age": ["min", "max"]
    }
)
````

````python
employee_df.groupby("Department").agg({
    "Salary": ["mean"],
    "Age": ["max"]
})
````

## GroupBy — The `observed` Parameter

### Learning Objective
Understand how `observed` controls whether Pandas includes unobserved categorical groups in the grouped result.

### 1. Understanding Categorical Data

A categorical column can have predefined categories, even if some categories do not appear in the actual data.

Consider this DataFrame:

```python
employee_data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Salary": [50000, 45000, 60000, 55000]
}

employee_df = pd.DataFrame(employee_data)

employee_df["Department"] = pd.Categorical(
    employee_df["Department"],
    categories=["IT", "HR", "Finance", "Marketing"]
)
```

Notice that `Marketing` is a predefined category, but no employee belongs to that department.

### 2. Default Behaviour: `observed=True`

In Pandas 3.0, `observed=True` is the default.

```python
employee_summary = employee_df.groupby("Department").agg(
    Average_Salary=("Salary", "mean")
)
```

Only categories that actually appear in the data are included in the result.

Therefore, `Marketing` will not appear.

### 3. Using `observed=False`

Setting `observed=False` includes all predefined categorical groups, even those without observations.

```python
employee_summary = employee_df.groupby(
    "Department",
    observed=False
).agg(
    Average_Salary=("Salary", "mean")
)
```

The result will also contain `Marketing`, but its average salary will be `NaN` because there are no employees in that category.

### Key Difference

| Parameter | Behaviour |
|---|---|
| `observed=True` | Includes only categories present in the data |
| `observed=False` | Includes all predefined categories, including unobserved ones |

**Remember:** `observed` is primarily relevant when grouping by categorical columns. It does not create new observations or modify your original DataFrame.

## GroupBy — The `get_group()` Method

### Learning Objective
Learn how to retrieve the rows belonging to a specific group from a GroupBy object.

### 1. Understanding `get_group()`

When you group a DataFrame, Pandas creates a GroupBy object.

You can use `.get_group()` to retrieve the original rows belonging to a particular group.

Using our existing `sales_df`:

```python
department_groups = sales_df.groupby("Department")

it_employees = department_groups.get_group("IT")

it_employees
```

This returns all rows where `Department` is `"IT"`.

### 2. Retrieving Another Group

You can retrieve any existing group by passing its group key:

```python
finance_employees = department_groups.get_group("Finance")

finance_employees
```

### 3. Important Points

- `.get_group()` returns a DataFrame containing the original rows of the selected group.
- It does not perform aggregation.
- It does not modify the original DataFrame.
- If the requested group does not exist, Pandas raises a `KeyError`.

### `get_group()` vs Aggregation

| Method | Purpose |
|---|---|
| `get_group()` | Retrieves the original rows of a specific group |
| `agg()` | Calculates summary statistics for groups |
| `transform()` | Returns group-level calculations aligned with original rows |
| `filter()` | Retains or removes entire groups based on a condition |
