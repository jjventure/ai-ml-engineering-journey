# Pandas — Filtering, Sorting and Column Updates

**Scope:** Exercises 5–10: multi-condition filtering, sorting, iloc selection, adding columns, updating values and conditional columns.

```python
import pandas as pd
```

```python
# Create a dictionary containing the employee data 
employee_data = {
    "Name": ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    "Age": [25, 30, 28, 35, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [50000, 45000, 60000, 55000, 48000]
}
```

```python
# Create a DataFrame from the dictionary
employee_df = pd.DataFrame(employee_data)
```

### Exercise 5 — Filtering with Multiple Conditions

**Objective:** Filter a DataFrame using multiple conditions.

Using the existing `employee_df` DataFrame, display only the employees who satisfy **both** of the following conditions:

1. Belong to the `IT` department.
2. Have a salary greater than ₹50,000.

**Requirements:**

* Use Boolean conditions to filter the DataFrame.
* Combine the conditions using the appropriate Pandas operator.
* Keep all four columns in the result.
* Do not modify the original DataFrame.

**Expected result:** Only employees satisfying both conditions should appear.


```python
employee_df
```

```python
# Filter employees based on department and salary
employee_df[(employee_df["Department"]=="IT") & (employee_df["Salary"]>50000)]
```

### Exercise 8 — Adding a New Column

The bonus for each employee should be **10% of their salary**.

**Requirements:**

* Create the `Bonus` column using the existing `Salary` column.

### Exercise 9 — Updating an Existing Column

**Objective:** Modify an existing column using a calculation.

Using the existing `employee_df` DataFrame, increase every employee's salary by **5%**.

**Requirements:**

* Update the existing `Salary` column.
* Calculate the new salary using the current salary values.
* Do not create a separate column for the increased salary.
* Display the updated DataFrame.

**Expected result:** Every employee's salary should increase by 5%, while the other columns remain unchanged.

**Hint:** Increasing a value by 5% means adding 5% of the original value to it.

### Exercise 10 — Creating a Conditional Column

**Objective:** Create a new column based on a condition.

Using the existing `employee_df` DataFrame, create a column named `Salary_Level`.

Apply the following rules:

* If an employee's salary is greater than ₹50,000, assign `"High"`.
* Otherwise, assign `"Standard"`.

**Requirements:**

* Use the existing `Salary` column to determine the salary level.
* Create a new column named `Salary_Level`.
* Apply the condition to every employee.
* Display the updated DataFrame.

**Hint:** Investigate the Pandas `where()` method or NumPy's `where()` function. Choose one approach and try implementing it yourself.


```python
employee_df
```

```python
import numpy as np 

employee_df["Salary Level"] = np.where(
    employee_df["Salary"] > 50000,
    "High",
    "Standard"
)

employee_df
```
