### Exercise 14 — Selecting Data with `loc` and `iloc`

**Objective:** Understand the difference between label-based and integer-position-based indexing.

Create the following DataFrame:

| Index | Name    | Department | Salary |
| ----: | ------- | ---------- | -----: |
|   101 | Alice   | IT         |  50000 |
|   205 | Bob     | HR         |  45000 |
|   309 | Charlie | Finance    |  60000 |
|   412 | Diana   | IT         |  55000 |

**Tasks:**

1. Create the DataFrame using a Python dictionary.
2. Set the index to `101`, `205`, `309`, and `412`.
3. Use `loc` to select Alice's row using its index label.
4. Use `iloc` to select the first row using its integer position.
5. Use `loc` to select Alice and Bob using their index labels.
6. Use `iloc` to select the first two rows using integer positions.

**Requirements:**

* Use both `loc` and `iloc`.
* Do not reset the index.
* Display the results of each selection.

**Learning goal:** Understand that `loc` selects by index labels, whereas `iloc` selects by integer positions.


```python
# 1. Create the DataFrame using a Python dict 
employee_data = {
    "Index": [101, 205, 309, 412],
    "Name": ["Alice", "Bob", "Charlie", "Diana"],
    "Department": ["IT", "HR", "Finance", "IT"],
    "Salary": [50000, 45000, 60000, 55000]
}

employee_df = pd.DataFrame(employee_data)
```

```python
employee_df
```

```python
# 2. Set the index to Index column values
employee_df = employee_df.set_index("Index")
employee_df
```

```python
# 3. Use loc to select Alice's row using it's Index label 
employee_df.loc[101]
```

```python
# 4. Use iloc to select the first row using its integer position 
employee_df.iloc[0]
```

```python
# 5. Use loc to select Alice and Bob's rows using index labels 
print(employee_df.loc[[101, 205]])
```

```python
# 6. Use iloc to select the first two rows using integer positions
employee_df.iloc[[0, 1]]
```

### Exercise 15 — Selecting Rows and Columns Together

**Objective:** Select specific rows and columns using Pandas indexing.

Use the existing `employee_df` DataFrame.

**Tasks:**

1. Use `loc` to display only the `Name` and `Salary` columns for Alice.
2. Use `loc` to display the `Name` and `Department` columns for Alice and Bob.
3. Use `iloc` to display the first two rows and the first two columns.
4. Use `iloc` to display the last two rows and the last two columns.

**Requirements:**

* Use `loc` and `iloc` as specified.
* Use a single indexing operation for each task.
* Do not modify the original DataFrame.

**Learning goal:** Understand how to select specific rows and columns simultaneously.


```python
employee_df
```

```python
# 1. Use loc to display only the Name and Salary columns for Alice 
employee_df.loc[101, ["Name", "Salary"]]
```

```python
# 2. Select the Name and Department columns for both Alice and Bob using .loc 
employee_df.loc[[101, 205], ["Name", "Department"]]
```

```python
# 3. Select the first two rows and the first two columns of employee_df using .iloc.
employee_df.iloc[0:2, 0:2]
```

```python
# Use .iloc to select the last two rows and the last two columns of employee_df.
employee_df.iloc[-2:, -2:]
```
