**Objective:** Filter rows based on whether a column's values match any value in a specified list.


Filter employees who work in either the `IT` or `Finance` departments.

**Your task:** Write the filtering code independently. Test it and inspect the output before moving to the next task.


````python
import pandas as pd

employee_data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Age": [25, 30, 28, 35, 26, 32],
    "Salary": [50000, 45000, 60000, 55000, 48000, 65000]
}

employee_df = pd.DataFrame(employee_data)
````

### Task 2 — Sorting with Three Columns

**Objective:** Sort employees using three different columns and sorting priorities.

Sort the employees according to these rules:

1. `Department` in ascending order (A–Z).
2. `Age` in descending order (oldest to youngest).
3. `Salary` in ascending order (lowest to highest).

**Requirements:**

### Task 3 — Understanding `inplace`

**Objective:** Understand the difference between returning a sorted DataFrame and modifying the original DataFrame.

**Tasks:**

1. Sort `employee_df` by `Salary` in descending order and store the result in a new variable called `sorted_employees`.
2. Display `sorted_employees`.
3. Display the original `employee_df` and verify whether its order has changed.

**Requirements:**

* Use `.sort_values()`.
* Do not modify the original DataFrame.
* Store the sorted result in a separate variable.

**Learning goal:** Understand how Pandas sorting operations behave when assigning their results to new variables.


````python
# by = ["Salary"]
# ascending = [False]

sorted_employees = employee_df.sort_values(
    by = "Salary",
    ascending = False,
)

print(f"sorted_employees: \n\n{sorted_employees}\n\n")
print(f"employee_df: \n\n{employee_df}")
````

### Task 4 — Resetting the Index After Sorting

**Objective:** Sort employees by salary and reset the DataFrame index.

**Tasks:**

1. Sort employees by `Salary` in descending order.
2. Store the sorted result in a variable called `sorted_employees`.
3. Reset the index so that it starts from 0 and follows sequential order.
4. Ensure the old index is not retained as a separate column.

**Requirements:**

* Use `.sort_values()`.
* Use `.reset_index()`.
* Do not modify the original DataFrame.

**Learning goal:** Understand how sorting preserves existing index labels and how resetting the index creates a new sequential index.


````python
sorted_employees = employee_df.sort_values(
    by = 'Salary',
    ascending = False
).reset_index(drop=True)

sorted_employees
````

### Task 5 — Final Sorting Challenge

**Objective:** Sort employees using multiple criteria and generate a clean sequential index.

**Sorting requirements:**

1. Department in ascending order (A–Z).
2. Salary in descending order (highest to lowest).

**Additional requirements:**

3. Store the result in `final_sorted_employees`.
4. Reset the index to start from 0.
5. Discard the original index.
6. Do not modify the original `employee_df`.

**Learning goal:** Independently combine multi-column sorting, sorting direction, variable assignment and index resetting.

**Your task:** Write the complete solution without referring to previous code.


````python
# by = ["Department", "Salary"]
# ascending = [True, False]

final_sorted_employees = employee_df.sort_values(
    by = ["Department", "Salary"],
    ascending = [True, False]
).reset_index(drop=True)

print(f"final_sorted_employees: \n\n{final_sorted_employees}\n\n")
print(f"employee_df: \n\n{employee_df}")
````
